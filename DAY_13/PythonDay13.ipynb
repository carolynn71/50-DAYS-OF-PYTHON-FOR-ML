### `break` Statement

The `break` statement is used to terminate the loop entirely. When `break` is encountered inside a loop, the loop is immediately exited, and program control resumes at the statement immediately following the loop.

```python
# Example of 'break'
print("Demonstrating 'break':")
for i in range(1, 10):
    if i == 5:
        break  # Exit the loop when i is 5
    print(i)
print("Loop finished (break encountered).")
```

### `continue` Statement

The `continue` statement is used to skip the rest of the code inside a loop for the current iteration only. When `continue` is encountered, the loop's next iteration begins without executing the statements after `continue` in the current iteration.

```python
# Example of 'continue'
print("\nDemonstrating 'continue':")
for i in range(1, 6):
    if i == 3:
        continue  # Skip printing when i is 3
    print(i)
print("Loop finished (continue encountered).")
```

### `pass` Statement

The `pass` statement is a null operation; nothing happens when it is executed. It is used as a placeholder where a statement is syntactically required but you don't want any command or code to execute. It can be useful when you are creating new code and want to leave a placeholder for future implementation without causing a syntax error.

```python
# Example of 'pass'
print("\nDemonstrating 'pass':")
for i in range(1, 4):
    if i == 2:
        pass  # Do nothing when i is 2
    print(i)

def my_function():
    pass # Placeholder for future implementation

print("Loop finished (pass encountered).")
print("my_function() is defined but does nothing.")
```
