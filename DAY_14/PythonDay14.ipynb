```python
# Python Built-in Functions: Theory and Code Examples

# Theory:
# Built-in functions are pre-defined functions that are always available for use in Python.
# You don't need to import them from any module; they are part of the core language.
# They perform common tasks and help make your code more concise and readable.
# Examples include functions for type conversion, mathematical operations, input/output, and more.

# --- Code Examples ---

# 1. print(): Used to display output to the console.
print("Hello, world!")
print("The answer is:", 42)

# 2. len(): Returns the number of items in an object (e.g., string, list, tuple).
my_string = "Python"
my_list = [1, 2, 3, 4, 5]
print(f"Length of string '{my_string}': {len(my_string)}")
print(f"Length of list {my_list}: {len(my_list)}")

# 3. type(): Returns the type of an object.
num = 10
text = "example"
print(f"Type of {num}: {type(num)}")
print(f"Type of '{text}': {type(text)}")

# 4. int(), float(), str(): Type conversion functions.
str_num = "123"
int_val = int(str_num)
print(f"Converted '{str_num}' to integer: {int_val} (type: {type(int_val)})\n")

float_num = 3.14
str_val = str(float_num)
print(f"Converted {float_num} to string: '{str_val}' (type: {type(str_val)})\n")

# 5. input(): Used to get input from the user.
# user_name = input("Enter your name: ")
# print(f"Hello, {user_name}!")

# 6. max(), min(): Return the largest and smallest item in an iterable or between two or more arguments.
numbers = [10, 5, 20, 15]
print(f"Max of {numbers}: {max(numbers)}")
print(f"Min of {numbers}: {min(numbers)}")
print(f"Max of (5, 12, 3): {max(5, 12, 3)}")

# 7. sum(): Returns the sum of all items in an iterable.
print(f"Sum of {numbers}: {sum(numbers)}")

# 8. abs(): Returns the absolute value of a number.
neg_num = -7
print(f"Absolute value of {neg_num}: {abs(neg_num)}")

# 9. range(): Generates a sequence of numbers (often used in loops).
print("Numbers from 0 to 4 using range:")
for i in range(5):
    print(i, end=" ")
print("\n")

# 10. round(): Rounds a number to a given precision.
pi_val = 3.14159
print(f"Round {pi_val} to 2 decimal places: {round(pi_val, 2)}")
```
