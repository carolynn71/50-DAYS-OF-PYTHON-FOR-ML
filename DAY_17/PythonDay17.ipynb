Strings in Python are sequences of characters, and you can access individual characters or portions of a string using **indexing** and **slicing**.

### 1. String Indexing

**Indexing** allows you to access a single character in a string by its position. Python uses zero-based indexing, meaning the first character is at index `0`.

You* Positive indices start from the beginning of the string (0, 1, 2, ...)
* Negative indices start from the end of the string (-1 for the last character, -2 for the second to last, ...)

```python
# Example of indexing
my_string = "Python Programming"

# Accessing the first character (positive index)
first_char = my_string[0]
print(f"First character: {first_char}")

# Accessing the fifth character
fifth_char = my_string[4]
print(f"Fifth character: {fifth_char}")

# Accessing the last character (negative index)
last_char = my_string[-1]
print(f"Last character: {last_char}")

# Accessing the second to last character
second_to_last_char = my_string[-2]
print(f"Second to last character: {second_to_last_char}")
```

### 2. String Slicing

**Slicing** allows you to extract a portion (substring) of a string. The syntax is `[start:end:step]`:

*   `start`: The starting index of the slice (inclusive). If omitted, it defaults to the beginning of the string (`0`).
*   `end`: The ending index of the slice (exclusive). If omitted, it defaults to the end of the string.
*   `step`: The number of characters to skip. If omitted, it defaults to `1`.

```python
# Example of slicing
my_string = "Python Programming"

# Slice from index 0 up to (but not including) index 6
slice1 = my_string[0:6]
print(f"Slice [0:6]: '{slice1}'")

# Slice from index 7 to the end of the string
slice2 = my_string[7:]
print(f"Slice [7:]: '{slice2}'")

# Slice from the beginning to index 6
slice3 = my_string[:6]
print(f"Slice [:6]: '{slice3}'")

# Slice the entire string
slice4 = my_string[:]
print(f"Slice [:]: '{slice4}'")

# Slice with negative indices (last 11 characters)
slice5 = my_string[-11:]
print(f"Slice [-11:]: '{slice5}'")

# Slice with a step (every second character)
slice6 = my_string[::2]
print(f"Slice [::2]: '{slice6}'")

# Reverse the string
reversed_string = my_string[::-1]
print(f"Reversed string [::-1]: '{reversed_string}'")
```
